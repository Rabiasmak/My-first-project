```python
import pandas as pd
full_data = pd.read_csv("/content/AccidentLondonBoroughs2223.csv")
```

```python

sutton_data = full_data[
    full_data['Local_Authority_Highway'] == 'E09000029'].copy()

print("Sutton dataset:", sutton_data.shape)
print("District code:", sutton_data['Local_Authority_District'].unique())

sutton_data.head()
```

```python
print("Data types detected by Python:")
print(sutton_data.dtypes)

sutton_data.info()
```

```python
# Reset sutton_data from full_data to ensure clean run if the cell is executed multiple times
sutton_data = full_data[
    full_data['Local_Authority_Highway'] == 'E09000029'
].copy()

# Keep these 7 variables as numerical
numeric_cols = [
    'Location_Easting_OSGR', 'Location_Northing_OSGR',
    'Longitude', 'Latitude', 'Number_of_Vehicles',
    'Number_of_Casualties', 'Speed_limit'
]

# Convert the other coded variables to categories
for col in sutton_data.columns:
    if col not in numeric_cols + ['Accident_Index', 'Date', 'Time']:
        sutton_data[col] = sutton_data[col].astype('category')

# Put accident severity in the correct order
sutton_data['Accident_Severity'] = sutton_data['Accident_Severity'].cat.reorder_categories(
    [3, 2, 1], ordered=True
)

# Correct date and time formats
sutton_data['Date'] = pd.to_datetime(sutton_data['Date'], dayfirst=True)
sutton_data['Time'] = pd.to_timedelta(sutton_data['Time'] + ':00')

# Display corrected data types
print(sutton_data.dtypes)
```

```python
print(sutton_data['Accident_Severity'].cat.ordered)
print(sutton_data['Accident_Severity'].cat.categories)
```

```python
# Select numerical variables (excluding timedelta columns which do not support skew/kurtosis)
numbers = sutton_data.select_dtypes(include='number').select_dtypes(exclude='timedelta')

# Calculate statistics
stats = numbers.describe().T
stats['Range'] = stats['max'] - stats['min']
stats['Skewness'] = numbers.skew()
stats['Kurtosis'] = numbers.kurt()

# Show results
display(stats.round(2))
```

```python

# Select categorical variables
categories = sutton_data.select_dtypes(include='category')

# Show frequencies, percentages and mode
for col in categories:
    print("\nVariable:", col)
    print("Mode:", categories[col].mode()[0])
    print("Frequency:")
    print(categories[col].value_counts(dropna=False))
    print("Percentage:")
    print((categories[col].value_counts(normalize=True, dropna=False) * 100).round(2))
```

```python

# Check the date range
print("First date:", sutton_data['Date'].min())
print("Last date:", sutton_data['Date'].max())

# Count accidents by year
years = sutton_data['Date'].dt.year.value_counts().sort_index()
print("\nAccidents by year:")
print(years)

# Count accidents by weekday
print("\nAccidents by weekday:")
print(sutton_data['Date'].dt.day_name().value_counts())

# Count accidents by hour
hours = (sutton_data['Time'].dt.total_seconds() // 3600).astype(int)
print("\nAccidents by hour:")
print(hours.value_counts().sort_index())
```

```python

# Check missing values in all columns
print("Missing values:")
print(sutton_data.isna().sum())

# Check duplicate accident IDs
print("\nDuplicate accidents:")
print(sutton_data['Accident_Index'].duplicated().sum())

# Check unusual -1 codes
print("\nSpecial -1 values:")
for col in categories:
    count = (categories[col] == -1).sum()
    if count > 0:
        print(col, count)

# Check date and weekday consistency
weekday = (sutton_data['Date'].dt.dayofweek + 1) % 7 + 1
print("\nWeekday mismatches:",
      (weekday != sutton_data['Day_of_Week'].astype(int)).sum())

# Check time is within 24 hours
print("Invalid time ranges:",
      ((sutton_data['Time'] < pd.Timedelta(0)) |
       (sutton_data['Time'] >= pd.Timedelta(days=1))).sum())
```

```python

# Check outliers using the IQR method
for col in numbers:
    Q1 = numbers[col].quantile(0.25)
    Q3 = numbers[col].quantile(0.75)
    IQR = Q3 - Q1

    outliers = (
        (numbers[col] < Q1 - 1.5 * IQR) |
        (numbers[col] > Q3 + 1.5 * IQR)
    ).sum()

    print(col, "- Potential outliers:", outliers)
```

```python

# Count accidents by severity
print("Severity counts:")
print(sutton_data['Accident_Severity'].value_counts())

# Calculate severity percentages
print("\nSeverity percentages:")
print((sutton_data['Accident_Severity'].value_counts(normalize=True) * 100).round(2))
```

```python

import matplotlib.pyplot as plt

# Graph 1: Accidents by year
years.plot(kind='bar', title='Accidents by Year')
plt.xlabel('Year')
plt.ylabel('Number of Accidents')
plt.show()

# Graph 2: Accidents by hour
hours.value_counts().sort_index().plot(
    kind='bar', title='Accidents by Hour'
)
plt.xlabel('Hour')
plt.ylabel('Number of Accidents')
plt.show()

# Graph 3: Accident severity
sutton_data['Accident_Severity'].value_counts().sort_index().plot(
    kind='bar', title='Accident Severity'
)
plt.xlabel('Severity (3=Slight, 2=Serious, 1=Fatal)')
plt.ylabel('Number of Accidents')
plt.show()
```

```python

# Check invalid numerical values
print("Invalid vehicle counts:",
      (sutton_data['Number_of_Vehicles'] < 1).sum())

print("Invalid casualty counts:",
      (sutton_data['Number_of_Casualties'] < 1).sum())

print("Invalid latitude:",
      (~sutton_data['Latitude'].between(-90, 90)).sum())

print("Invalid longitude:",
      (~sutton_data['Longitude'].between(-180, 180)).sum())

print("Speed limits:",
      sorted(sutton_data['Speed_limit'].unique()))

# Investigate -1 codes in junction-related variables
for col in ['Junction_Control', '2nd_Road_Class']:
    print("\n", col)
    print(pd.crosstab(
        sutton_data['Junction_Detail'] == 0,
        sutton_data[col] == -1
    ))

# Examine unusual carriageway hazard code
print("\nCarriageway hazard -1 records:")
display(sutton_data[
    sutton_data['Carriageway_Hazards'] == -1
][['Accident_Index', 'Carriageway_Hazards']])
```

```python

# Show the IQR limits for every numerical variable
for col in numbers:
    Q1 = numbers[col].quantile(0.25)
    Q3 = numbers[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    print(col)
    print("IQR:", round(IQR, 2))
    print("Lower limit:", round(lower, 2))
    print("Upper limit:", round(upper, 2))
    print()

# Examine accidents with unusually high counts
print("Highest vehicle counts:")
display(sutton_data.nlargest(
    5, 'Number_of_Vehicles'
)[['Accident_Index', 'Number_of_Vehicles']])

print("Highest casualty counts:")
display(sutton_data.nlargest(
    5, 'Number_of_Casualties'
)[['Accident_Index', 'Number_of_Casualties']])
```

```python

# Compare severity with important conditions
for col in ['Light_Conditions',
            'Road_Surface_Conditions',
            'Weather_Conditions']:

    print("\n", col)

    table = pd.crosstab(
        sutton_data[col],
        sutton_data['Accident_Severity']
    )

    table['Total'] = table.sum(axis=1)

    table['Serious/Fatal %'] = (
        (table[1] + table[2]) / table['Total'] * 100
    ).round(2)

    display(table)
```

```python
northing = sutton_data['Location_Northing_OSGR']

Q1 = northing.quantile(0.25)
Q3 = northing.quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

unusual = sutton_data[(northing < lower) | (northing > upper)]

display(unusual[[
    'Accident_Index',
    'Location_Northing_OSGR',
    'Latitude',
    'Longitude']])
```
